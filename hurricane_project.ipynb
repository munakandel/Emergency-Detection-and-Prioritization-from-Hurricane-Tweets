```python
import os, re, pandas as pd

path = os.path.expanduser("~/Desktop/community resiliance/useful.csv")
df = pd.read_csv(path)

df["tweet_clean"] = df["tweet"].astype(str)

def clean_text(s):
    s = s.lower()
    s = re.sub(r"http\S+|www\.\S+", " ", s)
    s = re.sub(r"@\w+", " ", s)
    s = s.replace("#", " ")
    s = re.sub(r"\s+", " ", s).strip()
    return s

df["tweet_clean"] = df["tweet_clean"].apply(clean_text)

# Save cleaned file
out = os.path.expanduser("~/Desktop/community resiliance/useful_clean.csv")
df.to_csv(out, index=False)

print("Cleaned file saved:", out)
print(df[["tweet","tweet_clean"]].head(10))
```

```python
import numpy as np
import pandas as pd

# Parse timestamps safely
df["_created_at"] = pd.to_datetime(df["_created_at"], errors="coerce")
df["_started_at"] = pd.to_datetime(df["_started_at"], errors="coerce")

# Use created_at as primary event time
event_time = df["_created_at"]

# Compute hours since event ONLY where time exists
max_time = event_time.max()
df["hours_since_event"] = (max_time - event_time).dt.total_seconds() / 3600

# Initialize recency_score as NaN
df["recency_score"] = np.nan

# Apply exponential decay ONLY to rows with valid timestamps
mask_valid_time = df["hours_since_event"].notna()
df.loc[mask_valid_time, "recency_score"] = np.exp(
    -0.1 * df.loc[mask_valid_time, "hours_since_event"]
)

# Clean up intermediate column
df.drop(columns=["hours_since_event"], inplace=True, errors="ignore")

# Sanity check
df["recency_score"].describe()
```

```python
import pandas as pd

# Make sure df exists first
df_model = df.copy()

print("df_model created. Shape:", df_model.shape)
```

```python
import numpy as np
import pandas as pd

def simplify_message_type(x):
    if pd.isna(x):
        return np.nan
    x = x.lower()
    if "not informative: personal" in x:
        return "personal"
    elif "informative" in x:
        return "informative"
    else:
        return "other"

# Create simplified label column
df_model["type_simple"] = df_model["type_of_message"].apply(simplify_message_type)

print(df_model["type_simple"].value_counts(dropna=False))
```

```python
train_df = df_model[
    df_model["type_simple"].notna() &
    (df_model["tweet_clean"].str.strip() != "")
].copy()

X_train = train_df["tweet_clean"].tolist()
y_train = train_df["type_simple"].tolist()

print("Training samples:", len(train_df))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

type_imputer = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=30000,
        ngram_range=(1, 2),
        stop_words="english",
        min_df=5
    )),
    ("clf", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        solver="lbfgs"
    ))
])

type_imputer.fit(X_train, y_train)
print("Model trained to learn message-type patterns")
```

```python
mask_missing = (
    df_model["type_simple"].isna() &
    (df_model["tweet_clean"].str.strip() != "")
)

X_missing = df_model.loc[mask_missing, "tweet_clean"].tolist()

print("Rows to impute:", len(X_missing))
```

```python
probs = type_imputer.predict_proba(X_missing)
preds = type_imputer.classes_[probs.argmax(axis=1)]
conf = probs.max(axis=1)
```

```python
CONF_THRESHOLD = 0.70  # safe choice

df_model.loc[mask_missing, "type_simple"] = np.where(
    conf >= CONF_THRESHOLD,
    preds,
    "other"
)

print(df_model["type_simple"].value_counts())
```

```python
# Create a final message type column
df_model["type_of_message_imputed"] = df_model["type_simple"]
```

```python
# Import transformer pipeline for LLM-based tweet classification
# and pandas/numpy for handling and storing tweet data and predictions
from transformers import pipeline
import pandas as pd
import numpy as np
```

```python
# Load a large pre-trained transformer (LLM) for zero-shot tweet classification..........................
# Used to automatically label tweets as emergency, support offer, or neutral without manual annotation
# zero_shot_classifier = pipeline(
#     "zero-shot-classification",
#     model="facebook/bart-large-mnli"
# )
zero_shot_classifier = pipeline(
    "zero-shot-classification",
    model="cross-encoder/nli-deberta-v3-small"
)
```

```python
# Define target categories for automatic tweet classification:
# emergency requests, support offers, and neutral (general information)
candidate_labels = ["emergency", "support offer", "neutral"]
```

```python
# Automatically classify each tweet into emergency, support offer, or neutral
    # using a pre-trained LLM (BART) and return both the predicted label and confidence score
def zero_shot_label(text):
    if not isinstance(text, str) or text.strip() == "":
        return None, None

    result = zero_shot_classifier(
        text,
        candidate_labels,
        hypothesis_template="This disaster tweet describes {}."
    )

    label = result["labels"][0] # Highest-confidence category
    score = result["scores"][0] # Confidence of the prediction

    return label, score
```

```python
# Automatically label each tweet as emergency, support offer, or neutral
# using a zero-shot LLM, and store both the predicted label and confidence score
df_model["auto_label"], df_model["auto_label_score"] = zip(*df_model["tweet_clean"].apply(zero_shot_label))
```

```python
# Check how many tweets were automatically labeled as emergency, support offer, or neutral by the LLM
df_model["auto_label"].value_counts()
```

```python
def finalize_label(row):
    label = row["auto_label"]
    score = row["auto_label_score"]

    # if label == "emergency" and score >= 0.40:.................................
    #     return "emergency"
    # elif label == "support offer" and score >= 0.50:
    #     return "support offer"
    # else:
    #     return "neutral"
    if label == "emergency" and score >= 0.60:
     return "emergency"
    elif label == "support offer" and score >= 0.50:
      return "support offer"
    else:
       return "neutral"
df_model["auto_label_final"] = df_model.apply(finalize_label, axis=1)

df_model["auto_label_final"].value_counts()
```

```python
import pandas as pd
import os

path = os.path.expanduser(
    "~/Desktop/community resiliance/final_modeling_dataset.csv"
)

df = pd.read_csv(path)

# Remove duplicate tweets for training
df_unique = df.drop_duplicates(subset="tweet_clean").copy()

print("Unique rows:", len(df_unique))
```

```python
label_map_task1 = {
    "emergency": 0,
    "support offer": 1,
    "neutral": 2
}

id2label_task1 = {v: k for k, v in label_map_task1.items()}

df_unique["label_task1"] = df_unique["auto_label_final"].map(label_map_task1)
```

```python
label_map_task2 = {
    "informative": 0,
    "personal": 1,
    "other": 2
}

id2label_task2 = {v: k for k, v in label_map_task2.items()}

df_unique["label_task2"] = df_unique["type_of_message_imputed"].map(label_map_task2)
```

```python
from cleanlab.classification import CleanLearning
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Auto-detect and remove mislabeled rows 
# X_text = df_unique["tweet_clean"].fillna("").tolist()
# y_t1   = df_unique["label_task1"].tolist()
import numpy as np
X_text = df_unique["tweet_clean"].fillna("").to_numpy()
y_t1   = df_unique["label_task1"].to_numpy()

base_pipe = SkPipeline([
    ("tfidf", TfidfVectorizer(max_features=20000, ngram_range=(1,2), min_df=3)),
    ("clf",   LogisticRegression(max_iter=1000, class_weight="balanced"))
])
cl = CleanLearning(base_pipe)
cl.fit(X_text, y_t1)
issues = cl.get_label_issues()

clean_mask = ~issues["is_label_issue"].values
df_unique  = df_unique[clean_mask].reset_index(drop=True)
df_unique = df_unique.dropna(subset=["label_task1", "label_task2"]).reset_index(drop=True)
print(f"CleanLab removed {(~clean_mask).sum()} noisy rows. Remaining: {len(df_unique)}")
```

```python
from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    df_unique,
    test_size=0.30,
    stratify=df_unique["label_task1"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label_task1"],
    random_state=42
)
```

```python
from datasets import Dataset

# Task 1 datasets
train_ds1 = Dataset.from_pandas(train_df[["tweet_clean", "label_task1"]])
val_ds1   = Dataset.from_pandas(val_df[["tweet_clean", "label_task1"]])
test_ds1  = Dataset.from_pandas(test_df[["tweet_clean", "label_task1"]])

# Task 2 datasets
train_ds2 = Dataset.from_pandas(train_df[["tweet_clean", "label_task2"]])
val_ds2   = Dataset.from_pandas(val_df[["tweet_clean", "label_task2"]])
test_ds2  = Dataset.from_pandas(test_df[["tweet_clean", "label_task2"]])
```

```python
from transformers import AutoTokenizer

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

# Rename function to avoid name conflict
def tokenize_function(batch):
    texts = [str(t) for t in batch["tweet_clean"]]
    
    return tokenizer(
        texts,
        truncation=True,
        padding="max_length",
        max_length=128
    )

# Apply tokenization
train_ds1 = train_ds1.map(tokenize_function, batched=True)
val_ds1   = val_ds1.map(tokenize_function, batched=True)
test_ds1  = test_ds1.map(tokenize_function, batched=True)

train_ds2 = train_ds2.map(tokenize_function, batched=True)
val_ds2   = val_ds2.map(tokenize_function, batched=True)
test_ds2  = test_ds2.map(tokenize_function, batched=True)
```

```python
# BERT 

from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
import numpy as np
from sklearn.metrics import accuracy_score, f1_score
```

```python

#  Clean text columns first 
train_df["tweet_clean"] = train_df["tweet_clean"].fillna("").astype(str)
val_df["tweet_clean"]   = val_df["tweet_clean"].fillna("").astype(str)
test_df["tweet_clean"]  = test_df["tweet_clean"].fillna("").astype(str)
```

```python
# Build HuggingFace datasets 

train_ds_bert_t1 = Dataset.from_pandas(train_df[["tweet_clean", "label_task1"]])
val_ds_bert_t1   = Dataset.from_pandas(val_df[["tweet_clean", "label_task1"]])
test_ds_bert_t1  = Dataset.from_pandas(test_df[["tweet_clean", "label_task1"]])
```

```python
# Load tokenizer 
# %%
# tokenizer
tokenizer_bert = AutoTokenizer.from_pretrained("bert-base-uncased")

def tokenize_bert(batch):
    texts = [str(x) for x in batch["tweet_clean"]]
    return tokenizer_bert(
        texts,
        truncation=True,
        padding="max_length",
        max_length=128
    )

train_ds_bert_t1 = train_ds_bert_t1.map(tokenize_bert, batched=True)
val_ds_bert_t1   = val_ds_bert_t1.map(tokenize_bert, batched=True)
test_ds_bert_t1  = test_ds_bert_t1.map(tokenize_bert, batched=True)
```

```python
# Rename label column 

train_ds_bert_t1 = train_ds_bert_t1.rename_column("label_task1", "labels")
val_ds_bert_t1   = val_ds_bert_t1.rename_column("label_task1", "labels")
test_ds_bert_t1  = test_ds_bert_t1.rename_column("label_task1", "labels")

train_ds_bert_t1.set_format("torch")
val_ds_bert_t1.set_format("torch")
test_ds_bert_t1.set_format("torch")
```

```python

# Load model 
model_bert_t1 = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=3
)
```

```python

#  Metrics 
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)
    return {
        "accuracy": accuracy_score(labels, preds),
        "f1_macro": f1_score(labels, preds, average="macro")
    }
```

```python


# Training arguments 
training_args_bert_t1 = TrainingArguments(
    output_dir="./bert_task1",
    num_train_epochs=10,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    report_to="none"
)
```

```python
import os
import numpy as np
import pandas as pd
import re

# load dataset and keep unique tweets only
path = os.path.expanduser("~/Desktop/community resiliance/final_modeling_dataset.csv")
df_full   = pd.read_csv(path)
df_unique = df_full.drop_duplicates(subset="tweet_clean").copy().reset_index(drop=True)
print(f"Full dataset: {len(df_full)} | Unique tweets: {len(df_unique)}")

# show how many tweets fall in each Task1 + Task2 combination
print("\nTask 1 + Task 2 label combinations:")
print(pd.crosstab(df_unique["auto_label_final"],
                  df_unique["type_of_message_imputed"]))


# Task 2 label multiplier
# this is where training results are used to drive ranking
# personal = person is directly in danger -> highest boost
# informative = verified factual emergency content -> medium boost
# other = unclear context -> no boost
TASK2_MULTIPLIER = {
    "personal":    1.30,
    "informative": 1.15,
    "other":       1.00
}

def get_task2_multiplier(label):
    if not isinstance(label, str):
        return 1.00
    return TASK2_MULTIPLIER.get(label.lower().strip(), 1.00)

df_unique["task2_multiplier"] = df_unique["type_of_message_imputed"].apply(
    get_task2_multiplier
)


# tiered keyword scoring - three levels of severity
TIER1 = ["trapped", "dying", "sos", "stranded", "injured", "killed",
          "dead", "death", "collapse", "explosion", "gas leak", "victim",
          "casualties", "fatalities", "critical", "life threatening",
          "need help", "please help", "cant breathe", "going to die"]

TIER2 = ["rescue", "emergency", "evacuate", "evacuation", "flooding",
          "flooded", "fire", "blaze", "burning", "dangerous", "urgent",
          "shelter", "hospital", "destroyed", "swept", "downed", "surge",
          "without power", "contamination", "mandatory", "curfew",
          "people trapped", "homes destroyed", "missing person"]

TIER3 = ["flood", "damage", "outage", "wind", "power", "closed", "shut",
         "warning", "alert", "missing", "breaking", "disaster", "rain",
         "winds", "storm", "bridges", "tunnels", "homeless", "debris"]

def keyword_score_tiered(text):
    if not isinstance(text, str):
        return 0.0, 0
    t = text.lower()
    t1 = sum(1.0 for kw in TIER1 if kw in t)
    t2 = sum(0.6 for kw in TIER2 if kw in t)
    t3 = sum(0.3 for kw in TIER3 if kw in t)
    raw_hits = int(t1 > 0) + int(t2 > 0) + int(t3 > 0)
    return min((t1 + t2 + t3) / 5.0, 1.0), raw_hits

kw_results = df_unique["tweet_clean"].apply(keyword_score_tiered)
df_unique["keyword_score"] = kw_results.apply(lambda x: x[0])
df_unique["keyword_hits"]  = kw_results.apply(lambda x: x[1])


# first person distress patterns
# someone saying "we are trapped" or "help us" is more urgent
# than someone reporting news about others
PERSONAL_DISTRESS = [
    r"\bwe are trapped\b", r"\bi am trapped\b", r"\bwe need help\b",
    r"\bi need help\b",    r"\bplease rescue\b", r"\bsomeone help\b",
    r"\bwe are stuck\b",   r"\bi am stuck\b",    r"\bour house\b",
    r"\bour street\b",     r"\bmy family\b",     r"\bmy children\b",
    r"\bmy home\b",        r"\bwe have no\b",    r"\bi have no\b",
    r"\bwe are stranded\b",r"\bhelp us\b",       r"\bsave us\b",
    r"\bplease send\b"
]

def personal_distress_score(text):
    if not isinstance(text, str):
        return 0.0
    t = text.lower()
    hits = sum(1 for p in PERSONAL_DISTRESS if re.search(p, t))
    return min(hits / 2.0, 1.0)

df_unique["personal_distress_score"] = df_unique["tweet_clean"].apply(
    personal_distress_score
)


# content-validated retweet score
# problem found in data: many high-retweet tweets were jokes or
# foreign language posts with zero emergency keywords
# fix: only give full retweet credit if tweet has keyword hits
df_unique["retweet_count"] = df_unique["retweet_count"].fillna(1)

# convert retweet count to percentile rank (0 to 1)
df_unique["retweet_percentile"] = df_unique["retweet_count"].rank(pct=True)

# full retweet score only if tweet has at least one keyword hit
# otherwise reduce it by 80% to penalize irrelevant viral tweets
df_unique["retweet_validated"] = np.where(
    df_unique["keyword_hits"] >= 1,
    df_unique["retweet_percentile"],
    df_unique["retweet_percentile"] * 0.2
)


# Task 1 model confidence (RoBERTa - best model for emergency detection)
# only applied to tweets labeled as emergency
df_unique["model_conf_score"] = np.where(
    df_unique["auto_label_final"] == "emergency",
    df_unique["auto_label_score"].fillna(0),
    0.0
)


# base urgency score from four signals
# recency was removed - 426 out of 1214 tweets had no timestamp
# and valid timestamps all clustered at 0.09-0.10 (no useful difference)
W_CONF     = 0.35
W_KEYWORD  = 0.30
W_PERSONAL = 0.20
W_RETWEET  = 0.15

df_unique["base_urgency"] = (
    W_CONF     * df_unique["model_conf_score"] +
    W_KEYWORD  * df_unique["keyword_score"] +
    W_PERSONAL * df_unique["personal_distress_score"] +
    W_RETWEET  * df_unique["retweet_validated"]
)

# apply Task 2 multiplier on top of base score
# this is where BERT training result is used to boost ranking
# emergency + personal gets 1.30x, emergency + informative gets 1.15x
df_unique["urgency_score"] = (
    df_unique["base_urgency"] * df_unique["task2_multiplier"]
).round(4)


# keep only emergency tweets with model confidence above 0.70
# removes tweets the model was uncertain about
emergency_ranked = (
    df_unique[
        (df_unique["auto_label_final"] == "emergency") &
        (df_unique["auto_label_score"] >= 0.70)
    ]
    .copy()
    .sort_values("urgency_score", ascending=False)
    .reset_index(drop=True)
)
emergency_ranked.index += 1
emergency_ranked.index.name = "rank"


def urgency_tier(score):
    if score >= 0.60:   return "Critical"
    elif score >= 0.48: return "High"
    elif score >= 0.35: return "Moderate"
    else:               return "Low"

emergency_ranked["urgency_tier"] = emergency_ranked["urgency_score"].apply(
    urgency_tier
)


print(f"\nTotal unique emergency tweets ranked: {len(emergency_ranked)}")

print("\nTier distribution:")
print(emergency_ranked["urgency_tier"].value_counts())

print("\nScore distribution:")
print(emergency_ranked["urgency_score"].describe().round(4))

print("\nBreakdown by Task 2 label inside emergency tweets:")
print(emergency_ranked["type_of_message_imputed"].value_counts())

display_cols = [
    "tweet_clean",
    "urgency_score",
    "urgency_tier",
    "auto_label_final",
    "type_of_message_imputed",
    "task2_multiplier",
    "model_conf_score",
    "keyword_score",
    "personal_distress_score",
    "retweet_validated",
    "retweet_count"
]

print("\nTop 20 most urgent tweets:")
display(emergency_ranked[display_cols].head(20))

print("\nBottom 10 lowest urgency tweets (sanity check):")
display(emergency_ranked[display_cols].tail(10))


save_path = os.path.expanduser(
    "~/Desktop/community resiliance/emergency_urgency_final.csv"
)
emergency_ranked.reset_index().to_csv(save_path, index=False)
print(f"\nSaved: {save_path}")
```
